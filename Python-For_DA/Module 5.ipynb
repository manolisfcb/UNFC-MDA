# Pandas Series

## Creating a Series:

### Creating a Series from a List:

```python
import pandas as pd

# Creating a Series from a list
data_series = pd.Series([1, 2, 3, 4, 5])
print(data_series)
```

### Creating a Series from a Dictionary:

```python
import pandas as pd

# Creating a Series from a dictionary
data_dict = {'a': 10, 'b': 20, 'c': 30}
data_series_dict = pd.Series(data_dict)
print(data_series_dict)
```

## Series Labels (Index)

### Custom Index:

```python
import pandas as pd

# Creating a Series with a custom index
data_series = pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e'])
print(data_series)
```

### Accessing Elements Using Labels:

```python
import pandas as pd

# Creating a Series with a custom index
data_series = pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e'])

# Accessing elements using labels
print(data_series['a'])
print(data_series[['a', 'c', 'e']])
```

### Modifying Index:

```python
import pandas as pd

# Creating a Series
data_series = pd.Series([1, 2, 3, 4, 5])

# Modifying the index
data_series.index = ['a', 'b', 'c', 'd', 'e']
print(data_series)
```

### Inspecting a Series

```python
import pandas as pd

# Sample Series
data_series = pd.Series([1, 2, 3, 4, 5])

# Displaying the first few elements
print(data_series.head(2))

# Displaying the last few elements
print(data_series.tail(2))

# Generating descriptive statistics
print(data_series.describe())
```

## Converting Columns to DateTime

### 1.	Basic Conversion

```python
import pandas as pd

# Sample DataFrame
data = {
    'date_strings': ['2023-07-21', '2023-07-22', '2023-07-23']
}
df = pd.DataFrame(data)

# Convert to datetime
df['date'] = pd.to_datetime(df['date_strings'])
df, df.dtypes
```

### 2.	Custom Date Format

```python
data = {
    'date_strings': ['21/07/2023', '22/07/2023', '23/07/2023']
}
df = pd.DataFrame(data)

# Convert to datetime with custom format
df['date'] = pd.to_datetime(df['date_strings'], format='%d/%m/%Y')
df
```

### 3.	Handling Different Date Formats

```python
import pandas as pd
from dateutil import parser

data = {
    'date_strings': ['2023-07-21', '21/07/2023', 'July 21, 2023']
}
df = pd.DataFrame(data)

def parse_date(date_string):
    formats = ['%Y-%m-%d', '%d/%m/%Y', '%B %d, %Y']
    for fmt in formats:
        try:
            return pd.to_datetime(date_string, format=fmt)
        except (ValueError, TypeError):
            continue
    try:
        return parser.parse(date_string)
    except (ValueError, TypeError):
        return pd.NaT

# Apply the function to the date_strings column
df['date_times'] = df['date_strings'].apply(parse_date)

df
```

### 4.	Including Time Information

```python
data = {
    'datetime_strings': ['2023-07-21 14:30:00', '2023-07-22 16:45:00', '2023-07-23 09:00:00']
}
df = pd.DataFrame(data)

# Convert to datetime
df['datetime'] = pd.to_datetime(df['datetime_strings'])
df, df.dtypes
```

### 5.	Different Time Zones

```python
data = {
    'datetime_strings': ['2023-07-21 14:30:00', '2023-07-22 16:45:00', '2023-07-23 09:00:00']
}
df = pd.DataFrame(data)

# Convert to datetime with time zone information
df['datetime'] = pd.to_datetime(df['datetime_strings']).dt.tz_localize('UTC')
df
```

```python
# Convert to another time zone
df['datetime_est'] = df['datetime'].dt.tz_convert('US/Eastern')
df
```

### 6.	Handling Invalid Dates

```python
data = {
    'date_strings': ['2023-07-21', 'invalid-date', '2023-07-23']
}
df = pd.DataFrame(data)

# Convert to datetime with error handling
df['date'] = pd.to_datetime(df['date_strings'], errors='coerce')
df
```

### 7.	Out-of-Bounds Dates

```python
import pandas as pd
from datetime import datetime

data = {
    'date_strings': ['2023-07-21', '9999-12-30', '1677-09-20', '2262-04-12']
}
df = pd.DataFrame(data)

# Custom function to handle date conversion
def custom_date_parser(date_str):
    try:
        # Attempt to parse using pandas
        return pd.to_datetime(date_str, errors='raise')
    except (pd.errors.OutOfBoundsDatetime, ValueError):
        # Handle out of bounds dates
        try:
            return datetime.strptime(date_str, '%Y-%m-%d')
        except ValueError:
            return date_str  # Return as string if all parsing fails

# Apply the custom function
df['date'] = df['date_strings'].apply(custom_date_parser)

df
```

### 8.	Extracting Date and Time Components

```python
data = {
    'datetime_strings': ['2023-07-21 14:30:00', '2023-07-22 16:45:00', '2023-07-23 09:00:00']
}
df = pd.DataFrame(data)

# Convert to datetime
df['datetime'] = pd.to_datetime(df['datetime_strings'])

# Extract components
df['year'] = df['datetime'].dt.year
df['month'] = df['datetime'].dt.month
df['day'] = df['datetime'].dt.day
df['hour'] = df['datetime'].dt.hour
df['minute'] = df['datetime'].dt.minute
df['second'] = df['datetime'].dt.second

df
```

### 9.	Working with Date Ranges

```python
# Create a range of dates
date_range = pd.date_range(start='2023-07-21', end='2023-07-27', freq='D')
df = pd.DataFrame(date_range, columns=['date'])

df
```

## Removing Duplicates

```python
import pandas as pd

# Sample DataFrame with duplicate rows
data = {
    'Name': ['John', 'Jane', 'Tom', 'Lucy', 'John'],
    'Age': [28, 24, 35, 29, 28],
    'Salary': [70000, 80000, 120000, 90000, 70000]
}
df = pd.DataFrame(data)

# Removing duplicate rows
df_no_duplicates = df.drop_duplicates()
df_no_duplicates
```

```python
# Removing duplicate rows based on 'Name' column
df_no_duplicates_name = df.drop_duplicates(subset=['Name'])
df_no_duplicates_name
```

```python
# Removing all duplicates, keeping none
df_no_duplicates_all = df.drop_duplicates(keep=False)
df_no_duplicates_all
```
