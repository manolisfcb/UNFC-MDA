# Functions and Methods

## Functions

### Creating a Function

```python
# Defining a function
def greet():
    print("Hello, World!")

# Calling the function
greet()
# Output: Hello, World!
```

### Positional-Only and Keyword-Only Arguments

#### 1.	Positional-Only Arguments 

```python
# Defining a function with positional-only arguments
def add(a, b):
    return a + b

# Calling the function
result = add(3, 5)
print(result)  # Output: 8
```

#### 2.	Keyword-Only Arguments 

```python
# Defining a function with multiple keyword-only arguments
def display_info(age, city):
    print(f"Age: {age}, City: {city}")

# Calling the function with keyword arguments
display_info(city="New York", age=30)
# Output: Age: 30, City: New York
```

### 3.	Mixed Positional-Only and Keyword-Only Arguments 

```python
# Defining a function with positional-only and keyword-only arguments
def example(pos1, pos2, kw1, kw2):
    print(f"Positional: {pos1}, {pos2}; Keyword: {kw1}, {kw2}")

# Calling the function
example(1, 2, kw2="b", kw1="a")
# Output: Positional: 1, 2; Keyword: a, b
```

## Functions with Return Values

```python
# Defining a function with a return value
def add(a, b):
    return a + b

# Calling the function and storing the result
result = add(5, 3)
print(result)  # Output: 8
```

# Built-in Functions and Methods

## Built-in Functions

```python
# Using the built-in len() function to get the length of a string
length = len("Hello")
print(length)  # Output: 5

# Using the built-in type() function to get the type of a variable
var_type = type(5)
print(var_type)  # Output: <class 'int'>
```

## Built-in Methods

```python
# Using the built-in str.upper() method to convert a string to uppercase
message = "hello"
print(message.upper())  # Output: HELLO

# Using the built-in list.append() method to add an element to a list
numbers = [1, 2, 3]
numbers.append(4)
print(numbers)  # Output: [1, 2, 3, 4]
```

## Lambda Function

```python
# Defining a lambda function to add two numbers
add = lambda x, y: x + y

# Calling the lambda function
result = add(3, 5)
print(result)  # Output: 8
```

# Python Modules and Packages

## Modules

### Creating a Module

```python
# save as my_module.py
def greet(name):
    print(f"Hello, {name}!")

def add(a, b):
    return a + b
```

### Using a Module

```python
import my_module

my_module.greet("Alice")  # Output: Hello, Alice!
result = my_module.add(3, 5)
print(result)  # Output: 8
```

## Standard Library Modules

### math module

```python
import math

print(math.sqrt(16))  # Output: 4.0
print(math.pi)  # Output: 3.141592653589793
```

### datetime module

```python
import datetime

now = datetime.datetime.now()
print(now)  # Output: current date and time
```

### random module

```python
import random

print(random.randint(1, 10))  # Output: random integer between 1 and 10
```

## Frequently Used Modules and Packages

### NumPy

#### Creating an Array

```python
import numpy as np

# Creating a NumPy array
arr = np.array([1, 2, 3, 4, 5])
print(arr)  # Output: [1 2 3 4 5]
```

#### Performing Arithmetic Operations

```python
# Multiplying each element by 2
print(arr * 2)  # Output: [ 2  4  6  8 10]
```

### Pandas

#### Creating a DataFrame

```python
import pandas as pd

# Creating a DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie'], 'Age': [25, 30, 35]}
df = pd.DataFrame(data)
print(df)
# Output:
#       Name  Age
# 0    Alice   25
# 1      Bob   30
# 2  Charlie   35
```

#### Accessing a Column

```python
# Accessing the 'Name' column
print(df['Name'])
# Output:
# 0      Alice
# 1        Bob
# 2    Charlie
# Name: Name, dtype: object
```

### SciPy

```python
#!pip install scipy
from scipy import stats
import numpy as np

# Generating a random sample
data = np.random.normal(0, 1, 1000)

# Calculating mean and standard deviation
mean = np.mean(data)
std_dev = np.std(data)
print(f'Mean: {mean}, Standard Deviation: {std_dev}')
```

### Scikit-learn

```python
#!pip install scikit-learn
from sklearn.linear_model import LinearRegression
import numpy as np

# Creating data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 3, 5, 7, 11])

# Creating the model
model = LinearRegression()
model.fit(X, y)

# Making a prediction
predicted = model.predict(np.array([[6]]))
print(predicted)  # Output: [12.2.]
```
