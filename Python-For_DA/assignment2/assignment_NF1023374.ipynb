## **TASK 6 – Creating New Columns** 
Create:
1.	PopulationMillions = Population / 1_000_000
2.	GDPCategory = "Low" / "Medium" / "High" based on:
3.	Low: below median GDP
4.	Medium: between median and 75th percentile
5.	High: above 75th percentile

Then count how many rows fall into each category.


````python
df_sample["PopulationMillions"] = df_sample["Population"] / 1_000_000

# Method 1: Using built-in pandas functions pd.qcut
df_sample["GDPCategory"] = pd.qcut(df_sample["GDPperCapita"], q=[0, 0.5, 0.75, 1], labels=["Low", "Medium", "High"])
df_sample
````

````python
# Method 2: Using a custom function
def gdp_category(row):
    if row["GDPperCapita"] < df_sample["GDPperCapita"].median():
        return "Low"
    elif row["GDPperCapita"] <= df_sample["GDPperCapita"].quantile(0.75):
        return "Medium"
    else:
        return "High"

df_sample["GDPCategory_custom"] = df_sample.apply(gdp_category, axis=1)
df_sample
````

````python
# Method 3: Using a lambda function
df_sample["GDPCategory_lambda"] = df_sample.apply(
    lambda row: "Low" if row["GDPperCapita"] < df_sample["GDPperCapita"].median() 
    else "Mediun" if row["GDPperCapita"] <= df_sample["GDPperCapita"].quantile(0.75) 
    else "High", axis=1) 
df_sample
````

````python
df_sample[[ "GDPperCapita","GDPCategory", "GDPCategory_lambda"]].head(10)
````

````python
df_sample["GDPCategory_custom"].value_counts()
````

Half of the dataset fis the median and is labeled as “Low,” while the Medium and High groups are nearly the same size because they follow the 50–75% and 75–100% quantile ranges.  
`qcut` handles boundaries and ties automatically, keeping the groups balanced, unlike the lambda and custom versions that uses fixed rules. I preferred using `qcut` to avoid issues with balancing and edge cases.

## **TASK 10 – Compare 1990 vs 2020**
1.	Identify countries with data for both 1990 and 2020.
2.	Compute change in LifeExpectancy for those countries.
3.	List the top 3 countries with the greatest improvement.


````python
df_1990_2020 = df_sample[df_sample["Year"].isin([1990, 2020])]
df_1990_2020
````

````python
boths_years = df_1990_2020.pivot(index="Country", columns="Year", values="LifeExpectancy").dropna()
boths_years
````


To compare changes in **LifeExpectancy** between 1990 and 2020, we pivoted the dataset so that **countries become the index**, **years become columns**, and **LifeExpectancy** becomes the cell values.  
This structure makes it easy to directly calculate how much life expectancy increased over the 30-year period.

After filtering only the countries that have data for **both 1990 and 2020**, we compute the difference:


````python
change_life_exp = boths_years[2020] - boths_years[1990] 
change_life_exp
````

````python
bigest_change = change_life_exp.nlargest(3) 
bigest_change
````

````python
bigest_change_v2 = change_life_exp.sort_values(ascending=False).head(3) 
bigest_change_v2
````

The countries with the **greatest improvement** in life expectancy between 1990 and 2020 are:

- **Congo, Dem. Rep.**  
- **Iran, Islamic Rep.**  
- **Chad**  

These countries show the largest positive changes, indicating significant progress in health conditions, healthcare access, and overall living standards over the past decades.
