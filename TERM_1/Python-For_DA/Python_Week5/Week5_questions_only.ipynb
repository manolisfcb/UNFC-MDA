## Inspecting a DataFrame

We can check the structure of a DataFrame with:

````python
print(df.head())
print(df.info())
print(df.describe())
````

## Accessing Data

You can access columns like this:

## Identifying and Handling Missing Values

You can detect missing values:
- `df.isnull()` → shows missing values as True/False
- `df.isnull().sum()` → shows count of missing values per column

You can fill missing values:
- `df['Age'].fillna(value)` → fill missing Age values

````python
print(df.isnull())
print("\nMissing values per column:\n", df.isnull().sum())

# Example: fill missing Age with the median
df['Age'] = df['Age'].fillna(df['Age'].median())
print("\nAfter filling missing Age values:\n", df)
````

## Removing Duplicates

You can remove duplicated rows:
- `df.duplicated()` → shows duplicated rows as True/False
- `df.drop_duplicates()` → removes duplicated rows

````python
print("\nDuplicated rows:\n", df.duplicated())
df = df.drop_duplicates()
print("\nDataFrame after removing duplicates:\n", df)
````

## Basic Filtering

You can select rows by conditions:
- Example: select all female patients (`Gender == 'F'`)
- Example: select all patients older than 40

````python
females = df[df['Gender'] == 'F']
print("All female patients:\n", females)

older_patients = df[df['Age'] > 40]
print("\nPatients older than 40:\n", older_patients)
````

## Exception Handling in Python

Exception handling lets you catch and manage errors that would normally stop your program.

Basic structure:
```python
try:
    # code that might raise an error
except ExceptionType:
    # what to do if the error occurs


````python
def safe_divide(a, b):
    try:
        result = a / "b"
        return result
    except Exception as e:
        return e

print(safe_divide(10, 2))   # normal division
print(safe_divide(5, 0))    # triggers exception
````

## Common Exception Types in Python

Here are some of the most commonly encountered exceptions:

| Exception | Description |
|----------|-------------|
| `ZeroDivisionError` | Raised when dividing by zero |
| `ValueError` | Raised when a function receives an argument of correct type but invalid value |
| `TypeError` | Raised when an operation is applied to an object of inappropriate type |
| `IndexError` | Raised when accessing an invalid index of a list or sequence |
| `KeyError` | Raised when a dictionary key is not found |
| `FileNotFoundError` | Raised when trying to open a non-existing file |

### Example: `ValueError`:

````python
def convert_to_int(value):
    try:
        return int(value)
    except ValueError:
        return "Cannot convert to integer."

print(convert_to_int("25"))      # Works fine
print(convert_to_int("hello"))   # Raises ValueError, caught by exception
````

### You may need to install folium once by running:

````python
import folium

def simple_folium_year_map(latitudes, longitudes, years):
    """
    Very simple helper function for MSc students.
    Plots points on Folium map, grouped by year with layer control.
    Different colors for each year.
    """
    # Define a simple list of colors
    colors = ['red', 'blue', 'green', 'orange', 'purple', 'black', 'brown']
    
    # Create base map at mean location
    m = folium.Map(location=[latitudes.mean(), longitudes.mean()], zoom_start=6)
    
    unique_years = sorted(years.dropna().unique())
    
    # Loop over unique years
    for idx, year in enumerate(unique_years):
        year_points = (years == year)
        fg = folium.FeatureGroup(name=str(int(year)))
        
        # Pick color (cycle if more years than colors)
        color = colors[idx % len(colors)]
        
        # Add all points for that year
        for lat, lon in zip(latitudes[year_points], longitudes[year_points]):
            folium.CircleMarker(
                location=[lat, lon],
                radius=3,
                color=color,
                fill=True,
                fill_opacity=0.7
            ).add_to(fg)
        
        fg.add_to(m)

    folium.LayerControl().add_to(m)
    return m
````
